```python
df.drop('parental_education_level',axis=1)#  drop  [ parental_education_level ]
```

```python
df.dropna(inplace=True)
```

```python
df.duplicated().sum()
```

```python
from sklearn.preprocessing import LabelEncoder
le= LabelEncoder()
cols=['student_id','gender','part_time_job','diet_quality','parental_education_level','extracurricular_participation','internet_quality']
for i in cols:
  df[i]=le.fit_transform(df[i])
```

```python
df.info()
```

```python
x=df.drop('exam_score',axis=1)
y=df['exam_score']
```

```python
from sklearn.model_selection import train_test_split
# train_test_split()
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.20, random_state=42)
```

```python
from sklearn.metrics import r2_score
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
X_train_scaled =sc.fit_transform(X_train)
X_test_scaled =sc.transform(X_test)
```

```python
#from sklearn.metrics import r2_score#,mean_absolute_error,mean_squared_error
# def evaluate(model,X_train, X_test, y_train, y_test):
#     train_pred =model.predict(X_train)
#     test_pred =model.predict(X_test)
#     print(f"Accuracy for your {model} is:")
#     print(f"accuracy_acore for training {model} is {r2_score(y_train,train_pred)}")
#     print(f"r2_score for testing {model} is {r2_score(y_test,test_pred)}")
```

```python
from sklearn.linear_model import LinearRegression
# le_model = LinearRegression()
# le_model.fit(X_train_scaled,y_train)
lr = LinearRegression()
lr.fit(X_train, y_train)

y_pred_lr = lr.predict(X_test)
y_pred_lr
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import Ridge
ridge = Ridge()
```

```python
ridge_param={
    'alpha':[0.1,1,5,10,50,100]
}
rigid_search = GridSearchCV(ridge,ridge_param,cv=5)
rigid_search.fit(X_train_scaled,y_train)
```

```python
rigid_search= rigid_search.predict(X_test)
rigid_search
```

```python
from sklearn.linear_model import Lasso
model_lasso = Lasso(alpha=0.1)
lasso_param = {
    'alpha':[0,1,2,3,5,10,15,20]
}
```

```python
lasso_search = GridSearchCV(model_lasso,lasso_param,cv=5)
lasso_search.fit(X_train_scaled,y_train)
```

```python
from sklearn.ensemble import RandomForestRegressor
rf = RandomForestRegressor(n_estimators=200)
rf.fit(X_train, y_train)

y_pred_rf = rf.predict(X_test)
y_pred_rf
```

```python
from sklearn.tree import DecisionTreeRegressor
dt=DecisionTreeRegressor(max_depth=5)
dt.fit(X_train, y_train)

y_pred_dt = dt.predict(X_test)
y_pred_dt
```

```python
from xgboost import XGBRegressor
model = XGBRegressor()
model.fit(X_train, y_train)

y_pred_xgb = model.predict(X_test)
y_pred_xgb
```

```python
models = {
    "Linear": LinearRegression(),
    "Ridge": Ridge(),
    "lasso": Lasso(),
    "RandomForest": RandomForestRegressor(),
    "DecisionTree": DecisionTreeRegressor(),
    "XGBoost": XGBRegressor()
}

for name, model in models.items():
    model.fit(X_train, y_train)
    pred = model.predict(X_test)
    print(name, r2_score(y_test, pred))
```
